# **Model Creation and Evaluation**

```python
profit_df.head()
```

```python
# Independent and dependent variable creation
X=profit_df[['RD_Spend','Administration_Spend','Marketing_Spend']] # X variable contains all inputs
y=profit_df['Profit'] # y is output
```

```python
# Traning and testing data creation
# 20% of the data goes to the test set
# 80% goes to the training set
# Every time you run the code, you get the same split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size= 0.2,random_state=4)
```

```python
X_test.shape,y_test.shape,X_train.shape,y_train.shape
```

```python
# Creating model
from sklearn.linear_model import LinearRegression
LR=LinearRegression() # object creation
LR.fit(X_train,y_train) # training of linear regression
y_predict=LR.predict(X_test)
```

```python
y_test # Actual values
```

```python
y_predict # Values predicted by model
```

```python
# Checking accuaracy with R square
from sklearn.metrics import r2_score
r2score=r2_score(y_test,y_predict)
```

```python
r2score
```

```python
# Testing with training data
y_train_predict=LR.predict(X_train)
```

```python
y_train
```

```python
y_train_predict
```

```python
# Checking accuarcy of train data
r2score2=r2_score(y_train,y_train_predict)
```

```python
r2score2     # Model is 94% accurate.
```

```python
# Ploting actual vs Predicted values
plt.figure(figsize=(10, 6))
plt.plot(y_train.values, label='Actual Output', marker='o')
plt.plot(y_train_predict, label='Predicted Output', marker='o')
plt.xlabel('Test Data Index')
plt.ylabel('Profit')
plt.title('Actual vs Predicted Profit')
plt.legend()
plt.grid(True)
plt.show()
```
